# Geometry and Database

- Import TPS (National)
- Sanity Check (Checking TPS code and compared it with Kemendagri Kelurahan Code)
- Merge join to geometry to TPS  dataset
- Exporting into new DKI CSV



```python
# reading dataframe
df = pd.read_csv('D:/Files/6. Work/3. Freelance/1. Datakota/2. Politico/politico-master/politico-master/data/PILPRES_2024-04-18.09-15.csv')
df.shape
```

```python
# Slicing the first 10 characters from each entry in the 'TPS' column
df['TPS_Sliced'] = df['tps'].str[:10]
df.head()
```

```python
# Sanity check of tps number using Kelurahan Ancol
count = (df['TPS_Sliced'] == '3172051003').sum()

# Print the count
print(count)
```

```python
# Aggregating df by TPS_Sliced column to get counts per Kelurahan
df_agg = df.groupby('TPS_Sliced').sum()

# Reset the index to have 'TPS_Sliced' as a column
df_agg = df_agg.reset_index()
df_agg
```

```python
#Creating DKI TPS csv file
df_dki = df_agg[df_agg['TPS_Sliced'].str.startswith('31')]

# Display the resulting DataFrame
df_dki
```

```python
# Import basemap
gdf = gpd.read_file('D:/Files/6. Work/3. Freelance/1. Datakota/2. Politico/politico-master/politico-master/data/DESA_KELURAHAN_BY_PROVINCE/DKI Jakarta_DESA_KELURAHAN.geojson')
gdf.head()
```

```python
# Create a new column 'DPT' in gdf without the dots
gdf['KODE_KEL_DESA_INT'] = gdf['KODE_KEL_DESA'].str.replace('.', '', regex=False)

# Ensure both columns are of type string and remove any leading/trailing spaces
gdf['KODE_KEL_DESA_INT'] = gdf['KODE_KEL_DESA_INT'].astype(str).str.replace(r'\D', '', regex=True).str.strip()

# Merging
gdf_dpt = pd.merge(gdf, df_agg, left_on='KODE_KEL_DESA_INT', right_on ='TPS_Sliced', how='left')
gdf_dpt
```

```python
gdf_dpt.info()
```

```python
#Exporting into polygon
output_file_path = 'D:/Files/6. Work/3. Freelance/1. Datakota/2. Politico/politico-master/politico-master/data/DKI.geojson'
gdf_dpt.to_file(output_file_path, driver='GeoJSON')
print(f"GeoDataFrame exported to {output_file_path}")
```
